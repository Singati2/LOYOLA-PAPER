# Minimal collision orders of degree-ratio histograms: certificates

Re-runs, in about two minutes on one CPU and well under 1 GB of memory, the exhaustive enumerations behind the minimal orders 13 (all trees) and 16 (maximum degree at most 4), the table of first orders by maximum degree, the early collisions of SDD and ISDD, and the witness trees with the all-order construction. Only `networkx` is needed (pinned to the version in `requirements-lock.txt`). No NIST data and nothing licensed is used. Set `REVISION` to the branch or tag to check.

```
import os, subprocess, sys
REVISION = os.environ.get('LOYOLA_REVISION', 'review/short-version')
if not os.path.exists('minimal.py'):
    if not os.path.exists('LOYOLA-PAPER'):
        subprocess.run(['git', 'clone', '--quiet', '--depth', '1', '--branch', REVISION,
                        'https://github.com/Singati2/LOYOLA-PAPER.git'], check=True)
    os.chdir('LOYOLA-PAPER/audit/certificates')
if 'google.colab' in sys.modules:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'networkx==3.6.1'], check=True)
import networkx; print('python', sys.version.split()[0], 'networkx', networkx.__version__)
```

```
def run(*args):
    out = subprocess.run([sys.executable, *args], capture_output=True, text=True, check=True).stdout
    print(out.strip().splitlines()[-1]); return out
a = run('minimal.py', '17')
assert a.strip().splitlines()[-1] == "{'all': 13, 'chem': 16}"
b = run('by_delta.py', '18')
assert "'4': 16, '5': 16" in b and "'None': 13" in b and "'3'" not in b.splitlines()[-1]
c = run('ratio_indices.py')
assert "('SDD', 'chem'): 9" in c and "('ISDD', 'chem'): 10" in c
d = run('witnesses.py')
assert d.count('OK for every order') == 2
print('ALL CERTIFICATES REPRODUCED')
```
